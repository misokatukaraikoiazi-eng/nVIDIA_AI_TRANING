```python
import numpy as np
from scipy.io import wavfile

# サンプリング周波数 16kHz（AI音声認識でよく使われる標準）
fs = 16000
t = np.linspace(0, 1, fs) # 1秒間

# 440Hz（ラの音）のサイン波を生成
data = np.sin(2 * np.pi * 440 * t)

# 16bit PCM形式で保存
data_int = (data * 32767).astype(np.int16)
wavfile.write("test_audio.wav", fs, data_int)

print("テスト音声 'test_audio.wav' を作成しました！")
```

```python
# Whisperで認識実行
result = model.transcribe("test_audio.wav")

# 結果を表示
print(f"【AIが聴き取った内容】\n{result['text']}")

```

```python
from ultralytics import YOLO

# 1. モデルの準備（メモリにあれば再ロードされません）
model = YOLO('yolov8n.pt')

# 2. 解析の実行（画像パスはご自身のものに変えてください）
# ここでは YOLO内蔵のサンプル画像を使います
img_path = 'https://ultralytics.com/images/bus.jpg'
results = model.predict(source=img_path)

# 3. 検知したデータの「中身」をのぞく
print("--- 検知結果のリスト ---")

for result in results:
    # 検出された各物体（boxes）をループで回す
    for box in result.boxes:
        # クラスID（何の物体か）を取得
        class_id = int(box.cls[0])
        label = model.names[class_id]

        # 確信度（AIがどれくらい自信があるか 0.0〜1.0）
        conf = float(box.conf[0])

        # 座標 (左上のx, y, 右下のx, y)
        coords = box.xyxy[0].tolist()

        # フォーマットして表示
        print(f"物体: {label:10} | 確信度: {conf:.2f} | 座標: {coords}")

print("-----------------------")
```

```python
from ultralytics import YOLO
from PIL import Image
import numpy as np

# 1. セグメンテーション用のモデル（-seg）をロード
# 学習済みデータも景色の分類が含まれているものを使います
model = YOLO('yolov8n-seg.pt')

# 2. 景色のサンプル画像を解析（YOLO内蔵のテスト画像）
# 砂浜と波、サーファーがいる景色です
img_path = 'https://ultralytics.com/images/living_room.jpg'
# 注意: living_room.jpg はサンプルですが、ソファ、椅子、鉢植えなどが写っています
# 景色のサンプルとしては、先ほどのバスの画像 (https://ultralytics.com/images/bus.jpg) を使うと、
# 人、バスの領域分割が見れます。

results = model.predict(source='https://ultralytics.com/images/bus.jpg', save=True)

# 3. 領域分割の結果（マスク）を表示
# 塗りつぶされた画像が出力されます
res_image = results[0].plot()
display(Image.fromarray(res_image[:, :, ::-1]))

# 4. 組み込み向け：1枚目の「マスクデータ（数値）」をのぞく
print("\n--- マスクデータの情報 ---")
if results[0].masks is not None:
    # 最初の検知物体のマスク（領域のxy座標）
    print(f"最初の物体の領域（座標リスト）:\n{results[0].masks.xy[0][:5]} ... (略)")
else:
    print("領域分割データはありません（検知されませんでした）")
```

```python
# 1. 必要なライブラリのインストールとインポート
!pip install ultralytics
from IPython.display import display, Javascript
from google.colab.output import eval_js
from base64 import b64decode
import numpy as np
import cv2
from ultralytics import YOLO

# モデルの準備（軽量なnモデル）
model = YOLO('yolov8n.pt')

# 2. JavaScriptでカメラを起動し、映像をPythonに送る関数
def take_photo(filename='photo.jpg', quality=0.8):
  js = Javascript('''
    async function takePhoto(quality) {
      const div = document.createElement('div');
      const video = document.createElement('video');
      video.style.display = 'block';
      const stream = await navigator.mediaDevices.getUserMedia({video: true});

      document.body.appendChild(div);
      div.appendChild(video);
      video.srcObject = stream;
      await video.play();

      // ここでループして解析し続ける簡易的な仕組み
      const canvas = document.createElement('canvas');
      canvas.width = video.videoWidth;
      canvas.height = video.videoHeight;

      // 1枚キャプチャして返す
      canvas.getContext('2d').drawImage(video, 0, 0);
      stream.getTracks().forEach(track => track.stop());
      div.remove();
      return canvas.toDataURL('image/jpeg', quality);
    }
    ''')
  display(js)
  data = eval_js('takePhoto({})'.format(quality))
  binary = b64decode(data.split(',')[1])
  with open(filename, 'wb') as f:
    f.write(binary)
  return filename

# 3. 実行：カメラで撮って、人を検知して表示
try:
  print("カメラを起動します...")
  filename = take_photo()

  # AIで検知（人だけを絞り込む場合は classes=[0]）
  results = model.predict(source=filename, classes=[0], conf=0.3)

  # 結果を表示
  from PIL import Image
  res_img = results[0].plot()
  display(Image.fromarray(res_img[:, :, ::-1]))

  # 座標データもデバッグ出力
  for box in results[0].boxes:
      print(f"人を検知しました！ 座標: {box.xyxy[0].tolist()}")

except Exception as err:
  print(str(err))
```
